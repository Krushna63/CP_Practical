```python
# student grade A - above 90 marks, B - above 80, C - above 70, D - above 60, F - below 60
# if elif else format
marks = int(input('Enter your marks: '))
if marks >= 90:
    grade = 'A'
elif marks >= 80:
    grade = 'B'
elif marks >= 70:
    grade = 'C'
elif marks >= 60:
    grade = 'D'
else:
    grade = 'F'
print('Your grade is:', grade)
```
